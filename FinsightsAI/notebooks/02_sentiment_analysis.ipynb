# Notebook 2: Sentiment Analysis Pipeline

This notebook:
- Applies VADER sentiment scoring to news headlines
- Aggregates sentiment to daily features
- Visualizes sentiment distributions and trends

```python
import sys
sys.path.insert(0, '..')
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.sentiment import add_vader_scores, aggregate_daily_sentiment, build_sentiment_dataset
sns.set_theme(style='darkgrid')
%matplotlib inline
```

```python
# Load preprocessed news
news = pd.read_csv('../data/processed/news_processed.csv')
print(news.shape)
news.head()
```

```python
# Apply VADER scoring
news_scored = add_vader_scores(news)
print(news_scored[['headline_clean','sentiment_label','sentiment_score','vader_compound']].head(10))
```

```python
# Correlation between VADER compound and labeled sentiment
corr = news_scored['vader_compound'].corr(news_scored['sentiment_score'])
print(f'Correlation (VADER compound vs label): {corr:.3f}')

fig, ax = plt.subplots(figsize=(8,5))
news_scored.boxplot(column='vader_compound', by='sentiment_label', ax=ax)
ax.set_title('VADER Compound Score by Sentiment Label')
plt.suptitle('')
plt.show()
```

```python
# Distribution of sentiment labels
fig, ax = plt.subplots(figsize=(7,4))
news_scored['sentiment_label'].value_counts().plot(kind='bar', ax=ax,
    color=['#00c88c','#94a3b8','#ff4d6d'])
ax.set_title('Sentiment Label Distribution')
ax.set_xlabel('Label')
ax.set_ylabel('Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Save scored news
news_scored.to_csv('../data/processed/news_scored.csv', index=False)
print('Saved news_scored.csv')
```
