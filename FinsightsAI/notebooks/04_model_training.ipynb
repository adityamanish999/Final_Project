# Notebook 4: Model Training & Evaluation

- Chronological train/val/test split
- Trains Logistic Regression, Random Forest, XGBoost, LightGBM
- Compares Accuracy, Precision, Recall, F1, ROC-AUC
- Plots ROC curves and confusion matrices

```python
import sys
sys.path.insert(0, '..')
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import RocCurveDisplay
from src.models import time_split, train_all, evaluate_all, save_all_models, get_best_model
from src.feature_engineer import ALL_FEATURES
sns.set_theme(style='darkgrid')
%matplotlib inline

TICKER = 'AAPL'
```

```python
# Load feature data
df = pd.read_csv(f'../data/processed/{TICKER}_features.csv', parse_dates=['Date'])
print(df.shape)
df.head()
```

```python
# Chronological split
X_train, y_train, X_val, y_val, X_test, y_test, dates_test = time_split(df)
print(f'Train: {len(X_train)} | Val: {len(X_val)} | Test: {len(X_test)}')
```

```python
# Train all models
fitted = train_all(X_train, y_train, X_val, y_val)
print('Models trained:', list(fitted.keys()))
```

```python
# Evaluate on test set
metrics_df = evaluate_all(fitted, X_test, y_test)
save_all_models(fitted, ticker=TICKER)
metrics_df.to_csv(f'../data/models/{TICKER}_metrics.csv')
print('\nTest Set Metrics:')
display(metrics_df)
```

```python
# ROC Curves
fig, ax = plt.subplots(figsize=(8, 6))
colors = ['#4f8ef7','#00c88c','#f59e0b','#a78bfa']
for (name, model), color in zip(fitted.items(), colors):
    RocCurveDisplay.from_estimator(model, X_test, y_test, ax=ax, name=name,
                                   color=color, alpha=0.85)
ax.plot([0,1],[0,1],'k--', label='Random')
ax.set_title(f'{TICKER} — ROC Curves (Test Set)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Confusion matrices
from sklearn.metrics import ConfusionMatrixDisplay
fig, axes = plt.subplots(1, 4, figsize=(18, 4))
for ax, (name, model) in zip(axes, fitted.items()):
    ConfusionMatrixDisplay.from_estimator(model, X_test, y_test, ax=ax,
                                          display_labels=['Down','Up'],
                                          colorbar=False)
    ax.set_title(name)
plt.tight_layout()
plt.show()
```

```python
# Best model
best_name, best_model = get_best_model(metrics_df, fitted)
print(f'Best model: {best_name}')
print(metrics_df.loc[best_name])
```
