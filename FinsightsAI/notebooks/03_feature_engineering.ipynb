# Notebook 3: Feature Engineering

- Computes all technical indicators (SMA, EMA, RSI, MACD, Bollinger Bands)
- Merges sentiment features
- Validates no data leakage (feature timestamps < target timestamps)

```python
import sys
sys.path.insert(0, '..')
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.feature_engineer import build_features, save_features, ALL_FEATURES
from src.sentiment import merge_sentiment_with_stock
from src.utils import PROC_DIR
sns.set_theme(style='darkgrid')
%matplotlib inline

TICKER = 'AAPL'
```

```python
# Load preprocessed stock data
stock = pd.read_csv(f'../data/processed/{TICKER}_processed.csv', parse_dates=['Date'])
print(stock.shape, stock.columns.tolist())
stock.head()
```

```python
# Build technical features (no sentiment for now)
features = build_features(stock, ticker=TICKER)
save_features(features, TICKER)
print(f'Features shape: {features.shape}')
features[['Date','Close','sma_20','ema_26','rsi','macd','bb_upper','bb_lower','target']].tail(10)
```

```python
# Validate no leakage: feature columns use data UP TO and including today
# Target uses TOMORROW's close. Let's verify the lag
print('Target distribution:')
print(features['target'].value_counts())
print('\nClass balance:', features['target'].mean().round(3))
```

```python
# Correlation heatmap (top features vs target)
corr_cols = ['daily_return','lag_1','lag_3','rsi','macd','bb_pct_b',
             'volatility_5','sma_5','target']
corr_cols = [c for c in corr_cols if c in features.columns]
corr = features[corr_cols].corr()

fig, ax = plt.subplots(figsize=(10,8))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdBu_r', center=0, ax=ax)
ax.set_title(f'{TICKER} — Feature Correlation Matrix')
plt.tight_layout()
plt.show()
```

```python
# Available features summary
available = [c for c in ALL_FEATURES if c in features.columns]
missing   = [c for c in ALL_FEATURES if c not in features.columns]
print(f'Available features ({len(available)}): {available}')
print(f'Missing (sentiment not merged yet): {missing}')
```
