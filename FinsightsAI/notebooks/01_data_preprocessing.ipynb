# Notebook 1: Data Collection & Preprocessing

This notebook covers:
- Fetching historical OHLCV data from Yahoo Finance
- Loading and cleaning the Kaggle financial news dataset
- Exploratory Data Analysis (EDA)

```python
import sys
sys.path.insert(0, '..')
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.data_loader import fetch_all_stocks, load_news_csv
from src.preprocessor import preprocess_stock, preprocess_news
sns.set_theme(style='darkgrid')
%matplotlib inline
```

```python
# --- 1. Fetch stock data ---
all_stocks = fetch_all_stocks()
print('Tickers fetched:', list(all_stocks.keys()))
aapl = all_stocks['AAPL']
aapl.head()
```

```python
# --- 2. EDA: Price trends ---
fig, axes = plt.subplots(2, 2, figsize=(16, 8))
for ax, (ticker, df) in zip(axes.flatten(), all_stocks.items()):
    ax.plot(pd.to_datetime(df['Date']), df['Close'])
    ax.set_title(f'{ticker} Close Price', fontsize=13)
    ax.set_xlabel('Date')
    ax.set_ylabel('Close ($)')
plt.tight_layout()
plt.show()
```

```python
# --- 3. Preprocess stock data ---
from src.preprocessor import preprocess_stock, save_processed_stock
processed = {}
for ticker, df in all_stocks.items():
    pdf = preprocess_stock(df, ticker)
    save_processed_stock(pdf, ticker)
    processed[ticker] = pdf
    print(f'{ticker}: {len(pdf)} rows | target balance: {pdf["target"].value_counts().to_dict()}')
```

```python
# --- 4. Load & preprocess news ---
# Change the path to where you downloaded the Kaggle CSV
NEWS_PATH = '../data/raw/all-data.csv'
raw_news = load_news_csv(NEWS_PATH)
print(raw_news.head())
print('\nSentiment distribution:')
print(raw_news['sentiment_label'].value_counts())
```

```python
from src.preprocessor import preprocess_news, save_processed_news
clean_news = preprocess_news(raw_news)
save_processed_news(clean_news)
clean_news.head()
```

```python
# --- 5. EDA: Missing values heatmap ---
fig, ax = plt.subplots(figsize=(10, 4))
sns.heatmap(processed['AAPL'].isnull(), cbar=False, ax=ax, yticklabels=False)
ax.set_title('Missing Values — AAPL Processed Data')
plt.show()
```
