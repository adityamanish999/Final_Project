# Notebook 5: SHAP Explainability

- Global feature importance (mean |SHAP|)
- Beeswarm plots showing direction of impact
- Local waterfall explanation for individual predictions
- Sentiment feature impact analysis

```python
import sys
sys.path.insert(0, '..')
import pandas as pd
import shap
import matplotlib.pyplot as plt
from src.models import time_split, load_model
from src.explainer import compute_shap_values, plot_global_importance, plot_waterfall
from src.feature_engineer import ALL_FEATURES
%matplotlib inline

TICKER     = 'AAPL'
MODEL_NAME = 'XGBoost'
```

```python
# Load data & model
df    = pd.read_csv(f'../data/processed/{TICKER}_features.csv', parse_dates=['Date'])
model = load_model(MODEL_NAME, ticker=TICKER)

X_train, y_train, X_val, y_val, X_test, y_test, dates_test = time_split(df)
available = [c for c in ALL_FEATURES if c in df.columns]
print(f'Features: {len(available)} | Test samples: {len(X_test)}')
```

```python
# Compute SHAP values on test set
shap_values = compute_shap_values(model, X_test, X_background=X_train)
print('SHAP values computed. Shape:', shap_values.values.shape)
```

```python
# Global importance plot
fig = plot_global_importance(shap_values, available, model_name=MODEL_NAME)
plt.show()
```

```python
# Local waterfall for the first test sample
fig = plot_waterfall(shap_values, sample_idx=0, feature_names=available,
                     model_name=MODEL_NAME)
plt.show()
print(f'Actual label: {y_test.iloc[0]} | Date: {dates_test.iloc[0].date()}')
```

```python
# Sentiment features impact
from src.feature_engineer import SENTIMENT_FEATURES
sent_available = [f for f in SENTIMENT_FEATURES if f in available]
if sent_available:
    feat_idx = [available.index(f) for f in sent_available]
    sv = shap_values.values
    if sv.ndim == 3:
        sv = sv[:, :, 1]
    sent_importance = pd.Series(
        abs(sv[:, feat_idx]).mean(axis=0),
        index=sent_available
    ).sort_values(ascending=True)

    fig, ax = plt.subplots(figsize=(8,4))
    sent_importance.plot(kind='barh', ax=ax, color='#4f8ef7')
    ax.set_title('Sentiment Feature SHAP Importance')
    ax.set_xlabel('Mean |SHAP value|')
    plt.tight_layout()
    plt.show()
else:
    print('No sentiment features found. Run the full pipeline with news data first.')
```
